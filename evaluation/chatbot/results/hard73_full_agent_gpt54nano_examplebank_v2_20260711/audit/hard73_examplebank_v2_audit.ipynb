# Auditoria hard-73 — example bank v2

Auditoria reproduzível das 49 falhas estritas do run `hard73_full_agent_gpt54nano_examplebank_v2_20260711`.


## Método

Reconciliamos os 73 registros com a adjudicação manual, comparamos o run anterior sem exemplos e verificamos sobreposição exata entre o banco curado e o hard-73.


```python
import json
from pathlib import Path

run = Path('evaluation/chatbot/results/hard73_full_agent_gpt54nano_examplebank_v2_20260711')
results = json.loads((run / 'variants/full_agent/results.json').read_text())
audit = json.loads((run / 'audit/failure_audit.json').read_text())
len(results['records']), len(audit['cases'])
```

```python
from collections import Counter

Counter(case['category'] for case in audit['cases'])
```

```python
failed_ids = {r['id'] for r in results['records'] if not r['result_match']}
audited_ids = {c['id'] for c in audit['cases']}
assert failed_ids == audited_ids
assert len(failed_ids) == 49
print('Reconciliation passed')
```

## Conclusão

O banco não contém perguntas ou SQLs exatamente iguais ao hard-73, mas algumas regras têm escopo excessivo e há um near-duplicate de template entre EX012 e GT215.
